```python
common_feature_mapping = {
    "Average Packet Size": "Avg Packet Size",
    "Total Length of Bwd Packets": "Bwd Packets Length Total",
    "Total Length of Fwd Packets": "Fwd Packets Length Total",
    "act_data_pkt_fwd": "Fwd Act Data Packets",
    "min_seg_size_forward": "Fwd Seg Size Min",
    "Init_Win_bytes_backward": "Init Bwd Win Bytes",
    "Init_Win_bytes_forward": "Init Fwd Win Bytes",
    "Max Packet Length": "Packet Length Max",
    "Min Packet Length": "Packet Length Min"
}

features_2017_for_cross = [
    col for col in df_2017_clean_names.columns
    if col not in {
        "Label",
        "Destination Port",
        "Fwd Header Length.1",
        "Bwd PSH Flags",
        "Bwd URG Flags",
        "Fwd Avg Bytes/Bulk",
        "Fwd Avg Packets/Bulk",
        "Fwd Avg Bulk Rate",
        "Bwd Avg Bytes/Bulk",
        "Bwd Avg Packets/Bulk",
        "Bwd Avg Bulk Rate"
    }
]

features_2018_for_cross = []

for feature in features_2017_for_cross:
    if feature in common_feature_mapping:
        features_2018_for_cross.append(common_feature_mapping[feature])
    elif feature in df_2018.columns:
        features_2018_for_cross.append(feature)

print("2017 features:", len(features_2017_for_cross))
print("2018 features:", len(features_2018_for_cross))
print("Feature counts match:", len(features_2017_for_cross) == len(features_2018_for_cross))
```

```python
import hashlib
import numpy as np

def create_row_hashes(df, feature_columns):
    values = df[feature_columns].replace([np.inf, -np.inf], np.nan)
    values = values.fillna(0)
    values = values.astype(np.float64)

    hashes = pd.util.hash_pandas_object(
        values,
        index=False
    )

    return hashes
```

```python
hashes_2017 = []

for file_path in files_2017:
    df = pd.read_csv(file_path)
    df.columns = df.columns.str.strip()

    for feature in features_2017_for_cross:
        if feature not in df.columns:
            raise ValueError(f"Missing feature in {file_path.name}: {feature}")

    row_hashes = create_row_hashes(df, features_2017_for_cross)

    hashes_2017.append(
        pd.DataFrame({
            "hash": row_hashes,
            "file": file_path.name
        })
    )

hashes_2017 = pd.concat(hashes_2017, ignore_index=True)

print("2017 hashed rows:", len(hashes_2017))
print("2017 unique hashes:", hashes_2017["hash"].nunique())
```

```python
hashes_2018 = []

for file_path in files_2018:
    df = pd.read_parquet(file_path)

    for feature in features_2018_for_cross:
        if feature not in df.columns:
            raise ValueError(f"Missing feature in {file_path.name}: {feature}")

    row_hashes = create_row_hashes(df, features_2018_for_cross)

    hashes_2018.append(
        pd.DataFrame({
            "hash": row_hashes,
            "file": file_path.name
        })
    )

hashes_2018 = pd.concat(hashes_2018, ignore_index=True)

print("2018 hashed rows:", len(hashes_2018))
print("2018 unique hashes:", hashes_2018["hash"].nunique())
```

```python
hashes_2017_unique = set(hashes_2017["hash"].unique())
hashes_2018_unique = set(hashes_2018["hash"].unique())

cross_dataset_duplicate_hashes = hashes_2017_unique.intersection(
    hashes_2018_unique
)

print(
    "Cross-dataset duplicate hashes:",
    len(cross_dataset_duplicate_hashes)
)
```

```python
cross_2017 = hashes_2017[
    hashes_2017["hash"].isin(cross_dataset_duplicate_hashes)
].copy()

cross_2018 = hashes_2018[
    hashes_2018["hash"].isin(cross_dataset_duplicate_hashes)
].copy()

print("2017 rows with cross-dataset duplicate fingerprints:", len(cross_2017))
print("2018 rows with cross-dataset duplicate fingerprints:", len(cross_2018))
```

```python
labels_by_hash_2017 = {}

for file_path in files_2017:
    df = pd.read_csv(file_path)
    df.columns = df.columns.str.strip()

    row_hashes = create_row_hashes(df, features_2017_for_cross)

    temp = pd.DataFrame({
        "hash": row_hashes,
        "label": df["Label"].values
    })

    temp = temp[temp["hash"].isin(cross_dataset_duplicate_hashes)]

    for hash_value, group in temp.groupby("hash"):
        labels_by_hash_2017[hash_value] = group["label"].unique().tolist()

print("2017 shared fingerprints with labels:", len(labels_by_hash_2017))
```

```python
labels_by_hash_2018 = {}

for file_path in files_2018:
    df = pd.read_parquet(file_path)

    row_hashes = create_row_hashes(df, features_2018_for_cross)

    temp = pd.DataFrame({
        "hash": row_hashes,
        "label": df["Label"].values
    })

    temp = temp[temp["hash"].isin(cross_dataset_duplicate_hashes)]

    for hash_value, group in temp.groupby("hash"):
        labels_by_hash_2018[hash_value] = group["label"].unique().tolist()

print("2018 shared fingerprints with labels:", len(labels_by_hash_2018))
```

```python
def to_binary_label(label):
    return "BENIGN" if str(label).strip().upper() == "BENIGN" else "ATTACK"

conflicting_binary_labels = []

for hash_value in cross_dataset_duplicate_hashes:
    labels_2017 = {
        to_binary_label(label)
        for label in labels_by_hash_2017[hash_value]
    }

    labels_2018 = {
        to_binary_label(label)
        for label in labels_by_hash_2018[hash_value]
    }

    if labels_2017 != labels_2018:
        conflicting_binary_labels.append({
            "hash": hash_value,
            "2017_labels": labels_2017,
            "2018_labels": labels_2018
        })

print("Cross-dataset fingerprints with conflicting binary labels:",
      len(conflicting_binary_labels))
```

```python
def create_row_hashes(df, feature_columns):
    values = df[feature_columns].astype(np.float64)

    hashes = pd.util.hash_pandas_object(
        values,
        index=False
    )

    return hashes
```

```python
hashes_2017 = pd.concat(
    [
        pd.DataFrame({
            "hash": create_row_hashes(
                pd.read_csv(file).rename(columns=lambda x: x.strip()),
                features_2017_for_cross
            ),
            "file": file.name
        })
        for file in files_2017
    ],
    ignore_index=True
)

hashes_2018 = pd.concat(
    [
        pd.DataFrame({
            "hash": create_row_hashes(
                pd.read_parquet(file),
                features_2018_for_cross
            ),
            "file": file.name
        })
        for file in files_2018
    ],
    ignore_index=True
)

print("2017 fingerprints:", len(hashes_2017))
print("2018 fingerprints:", len(hashes_2018))
```

```python
shared_hashes = set(hashes_2017["hash"]).intersection(
    set(hashes_2018["hash"])
)

print("Shared fingerprints:", len(shared_hashes))
```

```python
labels_by_hash_2017 = (
    pd.concat(
        [
            pd.DataFrame({
                "hash": create_row_hashes(
                    pd.read_csv(file).rename(columns=lambda x: x.strip()),
                    features_2017_for_cross
                ),
                "label": pd.read_csv(file)[" Label"].astype(str).str.strip()
            })
            for file in files_2017
        ],
        ignore_index=True
    )
)

labels_by_hash_2018 = (
    pd.concat(
        [
            pd.DataFrame({
                "hash": create_row_hashes(
                    pd.read_parquet(file),
                    features_2018_for_cross
                ),
                "label": pd.read_parquet(file)["Label"].astype(str).str.strip()
            })
            for file in files_2018
        ],
        ignore_index=True
    )
)

labels_by_hash_2017 = labels_by_hash_2017[
    labels_by_hash_2017["hash"].isin(shared_hashes)
]

labels_by_hash_2018 = labels_by_hash_2018[
    labels_by_hash_2018["hash"].isin(shared_hashes)
]

print("2017 shared rows:", len(labels_by_hash_2017))
print("2018 shared rows:", len(labels_by_hash_2018))
```

```python
def to_binary_label(label):
    return "BENIGN" if label.strip().upper() == "BENIGN" else "ATTACK"

labels_by_hash_2017["binary_label"] = labels_by_hash_2017["label"].map(to_binary_label)
labels_by_hash_2018["binary_label"] = labels_by_hash_2018["label"].map(to_binary_label)

binary_labels_2017 = (
    labels_by_hash_2017
    .groupby("hash")["binary_label"]
    .apply(set)
)

binary_labels_2018 = (
    labels_by_hash_2018
    .groupby("hash")["binary_label"]
    .apply(set)
)

conflicting_hashes = [
    h for h in shared_hashes
    if binary_labels_2017.get(h, set()) != binary_labels_2018.get(h, set())
]

print("Conflicting fingerprints:", len(conflicting_hashes))
```

```python
conflict_summary = pd.DataFrame({
    "hash": conflicting_hashes,
    "2017_labels": [binary_labels_2017[h] for h in conflicting_hashes],
    "2018_labels": [binary_labels_2018[h] for h in conflicting_hashes]
})

print(conflict_summary["2017_labels"].value_counts().to_string())
print()
print(conflict_summary["2018_labels"].value_counts().to_string())
```
